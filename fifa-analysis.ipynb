```python
# Log Value EUR and Log Wage EUR analysis

#Log_Value_Eur is showing the strong position relation with overall rating. As the correlation  is 0.90 which is highly correlated this might be the case of data leakage. 
#We are determining the features which will affect the overall rating but in this scenario Log_Value_Eur is affecting the overall rating positively but more than value affecting
#overall rating it seems to be overall rating which is affecting market value of any players which is data leakage in this scenario.
#Similar to Log_Wage_Eur, It shows 2 different pattern or trends. The Log_Wage_Eur above 9.5 is showing strong positive correlation whereas below 9.5 its showing no any relation.
#In below 9.5 the overall rating is not affected with wage value i.e Lower wage doesnt affects the players overall rating whereas higher wages are affecting the players overall raing.

#Both Log_Value_Eur and Log_Wage_Eur are likely results of Overall rating rather than causes of it.FIFA assigns market value and wage based on a players ability — meaning Overall 
#drives Value and Wage, not the other way around. Including these in a predictive model would cause data leakage and should be dropped during modelling phase.

#conclusion:
#These data leakage in modeling will create a problem where we are supposed to drop these columns in order to have greater accuracy.
```
